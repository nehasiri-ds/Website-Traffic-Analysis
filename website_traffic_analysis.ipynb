```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("website_traffic.csv")

df.head()
```

```python
print(df.info())
print(df.isnull().sum())
```

```python
total_sessions = df["Sessions"].sum()
total_users = df["Users"].sum()
avg_bounce = df["Bounce_Rate"].mean()
total_conversions = df["Conversions"].sum()

print("Total Sessions:", total_sessions)
print("Total Users:", total_users)
print("Average Bounce Rate:", round(avg_bounce,2))
print("Total Conversions:", total_conversions)
```

```python
source = df.groupby("Traffic_Source")["Sessions"].sum()

source.plot(kind="bar")
plt.title("Sessions by Traffic Source")
plt.savefig("traffic_source.png")
plt.show()
```

```python
page = df.groupby("Page")["Conversions"].sum()

page.plot(kind="bar")
plt.title("Conversions by Page")
plt.savefig("page_conversion.png")
plt.show()
```

```python
df["Date"] = pd.to_datetime(df["Date"])

trend = df.groupby("Date")["Sessions"].sum()

trend.plot()
plt.title("Daily Traffic")
plt.savefig("daily_trend.png")
plt.show()
```

Insights

1. Organic traffic contributes the highest number of sessions, indicating strong SEO performance.
2. Product and Pricing pages generate the highest conversions, showing users are more likely to convert when viewing these pages.
3. Blog page has the highest bounce rate, meaning users are leaving without further interaction.
4. Contact page shows the highest session duration and engagement, indicating user interest in reaching out.
5. Paid traffic contributes to conversions but has a relatively higher bounce rate, suggesting the need for better targeting.
6. Pages with higher session duration and pages per session show better user engagement and conversion potential.
7. Daily traffic trend shows consistent growth with slight variations, indicating stable website performance.
8. Direct traffic shows strong engagement, meaning returning or loyal users are interacting more effectively.
